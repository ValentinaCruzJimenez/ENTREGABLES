#**2. Limpieza de datos**

```python
# 2.1 Identificar las columnas con valores faltantes y aplica una estrategia de tratamiento adecuada.
# 2.2 Detectar y elimina los registros duplicados cuando corresponda.
# 2.3 Convertir las columnas que tengan tipos de datos incorrectos.
# 2.4 Identificar al menos dos variables categóricas que presenten inconsistencias y estandarízalas.
# 2.5 Después de realizar la limpieza, demuestra mediante código que las transformaciones fueron aplicadas correctamente.
df_animal = df_animal.drop(columns=['Animal code'], errors='ignore')

df_animal['Animal name'] = df_animal['Animal name'].fillna('Desconocido')

df_animal['Weight kg'] = df_animal['Weight kg'].fillna(df_animal['Weight kg'].median())
df_animal['Body Length cm'] = df_animal['Body Length cm'].fillna(df_animal['Body Length cm'].median())

df_animal = df_animal.drop_duplicates()

df_animal['Observation date'] = pd.to_datetime(df_animal['Observation date'], errors='coerce',dayfirst=True)

df_animal['Animal type'] = df_animal['Animal type'].replace({
    'red squirrell': 'red squirrel',
    'red squirle': 'red squirrel',
    'red squirel': 'red squirrel',
    'lynx?': 'lynx',
    'European bison™': 'European bison',
    'European bisson': 'European bison',
    'European buster': 'European bison',
    'ledgehod': 'hedgehog',
    'wedgehod': 'hedgehog'
})

df_animal['Country'] = df_animal['Country'].replace({
    'CC': 'Czech Republic',
    'CZ': 'Czech Republic',
    'Czech': 'Czech Republic',
    'DE': 'Germany',
    'HU': 'Hungary',
    'Hungry': 'Hungary',
    'PL': 'Poland'
})

df_animal['Animal type'] = df_animal['Animal type'].fillna('Desconocido')
df_animal['Country'] = df_animal['Country'].fillna('Desconocido')
df_animal['Gender'] = df_animal['Gender'].fillna('Desconocido')

df_animal['Observation date'] = df_animal['Observation date'].fillna(df_animal['Observation date'].mode()[0])
df_animal['Latitude'] = df_animal['Latitude'].fillna(df_animal['Latitude'].median())
df_animal['Longitude'] = df_animal['Longitude'].fillna(df_animal['Longitude'].median())

print(df_animal.isnull().sum())
```

#**3. NumPy**

```python
# 3.1 Crear una variable condicional utilizando np.where()
# Clasificamos a los animales como 'Grande' o 'Pequeño' basándonos en el peso promedio
peso_promedio = np.mean(df_animal['Weight kg'])
df_animal['Size'] = np.where(df_animal['Weight kg'] > peso_promedio, 'Grande', 'Pequeño')

# 3.2 Utilizar funciones estadísticas de NumPy
promedio_peso = np.mean(df_animal['Weight kg'])
desviacion_peso = np.std(df_animal['Weight kg'])
mediana_largo = np.median(df_animal['Body Length cm'])

print(f" Estadísticas Calculadas con NumPy ")
print(f"Peso promedio: {promedio_peso:.2f} kg")
print(f"Desviación estándar del peso: {desviacion_peso:.2f} kg")
print(f"Mediana de la longitud del cuerpo: {mediana_largo:.2f} cm")

print("\n Primeras filas con la nueva variable condicional ('Size') ")
display(df_animal[['Animal type', 'Weight kg', 'Size']].head(10))
```

```python
conteo_tamaños = df_animal['Size'].value_counts()
print(" Conteo de animales por Tamaño ")
print(conteo_tamaños)
# 3.3 Explica brevemente qué representa el resultado obtenido
# De acuerdo con el resultado se evidencia que hay más animales Pequeños (782) que grandes (60)
```

#**4. Análisis con groupby()**

```python
# 4.1 ¿Cuál es el número de observaciones por tipo de animal?
observaciones_por_tipo = df_animal.groupby('Animal type').size().reset_index(name='Observaciones')
display(observaciones_por_tipo)
```

```python
# 4.2 ¿Cuál es el promedio de una variable numérica según el tipo de animal?
promedio_peso_por_tipo = df_animal.groupby('Animal type')['Weight kg'].mean().reset_index(name='Peso Promedio (kg)')
display(promedio_peso_por_tipo)
```

```python
# 4.3 ¿Cuál es el promedio de una variable numérica según el país?
promedio_peso_por_pais = df_animal.groupby('Country')['Weight kg'].mean().reset_index(name='Peso Promedio (kg)')
display(promedio_peso_por_pais)
```
